```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print("Corrected temperature readings:", corrected)

fahrenheit = 9 / 5 * corrected + 32
print("Temperature in Fahrenheit:", fahrenheit)

greater_than_32 = corrected[corrected > 32]
print("Readings greater than 32°C:", greater_than_32)

count = np.sum(corrected > 32)
print("Number of readings greater than 32°C:", count)
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_index = np.argmax(steps)
position = np.unravel_index(max_index, steps.shape)

print("Maximum value:", steps[position])
print("Position:", position)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500

print("Original:", original)
print("Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)

print("Matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flattened:", flattened)
print("Raveled:", raveled)

raveled[0] = 100

print("Matrix after modifying ravel:")
print(matrix)
print("Raveled:", raveled)

flattened[1] = 200

print("Matrix after modifying flatten:")
print(matrix)
print("Flattened:", flattened)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

print("X.T:")
print(X.T)

xtx = X.T @ X

print("X.T @ X:")
print(xtx)

inverse = np.linalg.inv(xtx)

print("Inverse:")
print(inverse)

beta = inverse @ X.T @ y

print("Coefficients:", beta)

new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("Predicted assistance score:", prediction)
```
