```python
#Q1
roll = "1024170362"

L = [int(digit) * 10 for digit in roll]

#1
print("Original List:", L)

#2
L.append(55)
print("After append:", L)

L.insert(3, 99)
print("After insert:", L)

#3
L.remove(99)
print("After remove:", L)
L.pop()
print("After pop:", L)

#4
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

#5
print("First 3:", L[:3])
print("Last 3:", L[-3:])

#6
average = sum(L) / len(L)
greater = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average:", greater)
```

```python
#Q3
import random
from collections import Counter

random.seed(1024170362)

#1.
numbers = [random.randint(100,900) for _ in range(100)]

print(numbers)

#2.
odd = [x for x in numbers if x % 2 == 1]
print("Odd Count:", len(odd))

#3.
even = [x for x in numbers if x % 2 == 0]
print("Even Count:", len(even))


#4.
def prime(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

prime_list = [x for x in numbers if prime(x)]

print("Prime Count:", len(prime_list))
print("Prime Numbers:", prime_list)

#5.
count = Counter(numbers)
value, freq = count.most_common(1)[0]

print("Most Frequent:", value)
print("Frequency:", freq)
```

```python
#Q4
digits = [int(x) for x in roll]

A = {x*7 for x in digits}
B = {x*9 for x in digits}

print("A =", A)
print("B =", B)

#1.
print("Union:", A.union(B))

#2.
print("Intersection:", A.intersection(B))

#3.
print("A-B:", A.difference(B))
print("B-A:", B.difference(A))


#4.
print("Symmetric Difference:", A.symmetric_difference(B))

#5.
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

#6.
x = int(input("Enter value to remove: "))

A.discard(x)

print(A)
```
